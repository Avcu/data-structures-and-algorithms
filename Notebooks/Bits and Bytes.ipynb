## Playing with bits and bytes

#### Sometimes one has to go deep and deal with the stored bits and bytes rather than variables. Here, I have some functions that might be useful in such cases.

### Bin: returns the integer in binary format

```python
a = 30
a_bin = bin(a)

print("a in binary: {}".format(a_bin))
print("Type: {}".format(type(a_bin)))
```

```python
x = -3
x_bin = bin(x)

print("x in binary: {}".format(x_bin))
```

```python
print("x in binary: {}".format(bin(x & (0xFFFFFFFF))))
```

- 0x7FFFFFFF corresponds to max integer (01111...1111) --> 2**31-1
- 0xFFFFFFFF corresponds to min integer (11111...1111) --> 2**32

To convert positive integers to negative:
1. Invert all the bits
2. Add 1

For '3': 00..0011 -> 11..1100 -> 11..11101 ('-3')


#### Converting unsigned integer to Python's 32-bit pattern 

```python
mask = 0xFFFFFFFF
maxInt = 0x7FFFFFFF
res = 2**31 + 3
print(res)

if res > maxInt:
    # ~x = -(x+1)
    res = ~(res ^ mask)
print(res)
```

### `>>`: shifts the bits to the right


```python
b = a >> 1

b_bin = bin(b)

print("a: {} and b: {}".format(a, b))
print("a in binary: {}".format(a_bin))
print("b in binary: {}".format(b_bin))
```

### `<<`: shifts the bits to the left



```python
c = a << 1

c_bin = bin(c)

print("a: {} and c: {}".format(a, c))
print("a in binary: {}".format(a_bin))
print("c in binary: {}".format(c_bin))
```
